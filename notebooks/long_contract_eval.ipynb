```python
# 📦 Install necessary packages
!pip install -q transformers evaluate scikit-learn PyPDF2 nltk

import torch
import nltk
import numpy as np
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from PyPDF2 import PdfReader
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
import re

nltk.download("punkt")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ✅ Model paths (replace if needed)
model_paths = {
    "LegalBERT": "Kanishkagarwal6101/LegalBERT_LegalAnalyzer",
    "BERT": "Kanishkagarwal6101/BERT_LegalAnalyzer",
    "Longformer": "Kanishkagarwal6101/Longformer_LegalAnalyzer"
}

# ✅ Label maps
label2id = {
    "Business": 0, "Confidentiality": 1, "Consumers": 2, "Declarations": 3, "Economy": 4,
    "Education": 5, "Employment": 6, "Environment": 7, "External Relations": 8, "Fairness": 9,
    "Health": 10, "IP & Rights": 11, "Indemnification": 12, "Legal Governance": 13,
    "Miscellaneous": 14, "Payment": 15, "Social": 16, "Termination": 17
}
id2label = {v: k for k, v in label2id.items()}

# ✅ Load and smartly chunk Shuttle Contract
reader = PdfReader("/content/SampleContract-Shuttle.pdf")
text = "\n".join([page.extract_text() for page in reader.pages if page.extract_text()])

# Split based on numbered headings (1. ..., 2. ..., etc.)
clause_chunks = re.split(r"\n?\s*\d+\.\s+", text)
clause_chunks = [chunk.strip() for chunk in clause_chunks if chunk.strip()]
clause_chunks = clause_chunks[:24]  # ✅ Force limit to 24 important clauses
print(f"✅ Total clause-based chunks: {len(clause_chunks)}")

# ✅ Correct expected labels for these 24 clauses
expected_labels = [
    "Business", "Employment", "Payment", "Fairness", "Payment", "Payment",
    "Payment", "Fairness", "Fairness", "Fairness", "Miscellaneous",
    "Termination", "Indemnification", "Indemnification",
    "Fairness", "Fairness", "Termination", "Fairness",
    "Declarations", "Declarations", "Employment", "Fairness",
    "Fairness", "Miscellaneous"
]


true_ids = [label2id[label] for label in expected_labels]

# ✅ Loop through all models
models = {
    "LegalBERT_LegalAnalyzer": "Kanishkagarwal6101/LegalBERT_LegalAnalyzer",
    "BERT_LegalAnalyzer": "Kanishkagarwal6101/BERT_LegalAnalyzer",
    "Longformer_LegalAnalyzer": "Kanishkagarwal6101/Longformer_LegalAnalyzer"
}
for model_name, model_path in models.items():
    print(f"\n🔵 Evaluating model: {model_name}")

    # ✅ Load model + tokenizer
    tokenizer = AutoTokenizer.from_pretrained(model_path)
    model = AutoModelForSequenceClassification.from_pretrained(model_path).to(device)

    predicted_ids = []
    for clause in clause_chunks:
        if "longformer" in model_name.lower():
            max_len = 4096  # ✅ Longformer large context
        else:
            max_len = 512   # ✅ BERT and Legal-BERT standard context

        inputs = tokenizer(clause, return_tensors="pt", truncation=True, padding=True, max_length=max_len).to(device)
        with torch.no_grad():
            logits = model(**inputs).logits
            pred_id = torch.argmax(logits, dim=1).item()
        predicted_ids.append(pred_id)

    # ✅ Convert predictions to labels
    predicted_labels = [id2label[p] for p in predicted_ids]

    # ✅ Evaluate
    predicted_ids = predicted_ids[:len(true_ids)]
    acc = accuracy_score(true_ids, predicted_ids)
    prec, rec, f1, _ = precision_recall_fscore_support(true_ids, predicted_ids, average="weighted", zero_division=0)

    print(f"✅ Accuracy: {acc * 100:.2f}%")
    print(f"✅ Precision: {prec * 100:.2f}%")
    print(f"✅ Recall: {rec * 100:.2f}%")
    print(f"✅ F1 Score: {f1 * 100:.2f}%")

    # ✅ Save breakdown
    df = pd.DataFrame({
        "Clause": clause_chunks[:len(true_ids)],
        "Predicted": predicted_labels,
        "Expected": expected_labels
    })
    df.to_csv(f"{model_name}_shuttle_clause_eval.csv", index=False)
    print(f"📁 Saved evaluation to {model_name}_shuttle_clause_eval.csv")
```
